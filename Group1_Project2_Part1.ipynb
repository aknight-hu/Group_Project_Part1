```python
# Create a new column for total available bikes
df["TOTAL_BIKES_AVAILABLE"] = (
    df["NUM_BIKES_AVAILABLE"] + df["NUM_EBIKES_AVAILABLE"]
)
```

```python
# Display the first five rows
display(df[["NAME", "NUM_BIKES_AVAILABLE",
          "NUM_EBIKES_AVAILABLE",
          "TOTAL_BIKES_AVAILABLE"]].head())
```

We added the number of normal bikes and electric bikes available at each station to a new column I called TOTAL_BIKES_AVAILABLE. The total number of bikes that are available at each Capital Bikeshare station is shown in this column. It aids in our understanding of the number of bikes available for rent at each site.

```python
# Filter stations located in Washington, DC
dc_stations = df[df["REGION_NAME"] == "Washington, DC"]

# Display the first five results
display(dc_stations[["NAME", "REGION_NAME",
                   "TOTAL_BIKES_AVAILABLE"]].head())
```

We filtered the dataset to display only Capital Bikeshare locations in Washington, DC. A selection of 396 stations was produced as a result. This filter is interesting because it lets us concentrate on the availability of bikes in Washington, DC rather than examining every station in the dataset.

```python
# Filter stations with at least 20 bikes available
high_availability = df[df["TOTAL_BIKES_AVAILABLE"] >= 20]

# Display the first five results
display(high_availability[["NAME",
                         "TOTAL_BIKES_AVAILABLE"]].head())
```

We filtered the dataset in order to identify Capital Bikeshare locations with at least 20 bikes available. A selection of 64 stations was produced as a result. This filter is interesting because it makes it easier to find stations with lots of bikes available, which could be helpful for anyone wishing to rent a bike.

```python
# Idenitfy and select columns that have categorical data
categorical_columns = df.select_dtypes(
    include=["object", "str", "category"]
).columns

print("Categorical columns and their unique values:\n")

for column in categorical_columns:
    print(f"\n{column}:")
    print("Number of unique values:", df[column].nunique(dropna=False))
    print(df[column].value_counts(dropna=False).to_dict())
```

```python
# Show a shorter list of categories
important_categories = [
    "STATION_TYPE",
    "IS_INSTALLED",
    "IS_RETURNING",
    "IS_RENTING",
    "HAS_KIOSK",
    "REGION_NAME",
    "RENTAL_METHODS"
]

for column in important_categories:
    if column in df.columns:
        print(f"\n{column}")
        print(df[column].value_counts(dropna=False))
```

```python
 # Now we are going to group and aggregate the data
# First group stations by region and calculate the summary statistics
region_summary = df.groupby("REGION_NAME").agg(
    station_count=("NAME", "count"),
    average_capacity=("CAPACITY", "mean"),
    total_capacity=("CAPACITY", "sum"),
    average_bikes_available=("NUM_BIKES_AVAILABLE", "mean"),
    total_bikes_available=("NUM_BIKES_AVAILABLE", "sum"),
    average_ebikes_available=("NUM_EBIKES_AVAILABLE", "mean")
)
```

```python
#Round averages to two decimal places 
region_summary = region_summary.round(2)
```

```python
## Display regions with the most stations first
region_summary = region_summary.sort_values(
    by="station_count",
    ascending=False
)

region_summary
```

```python
# Visualize the number of stations by region/area
# Create a bar chart of station counts by region
region_summary["station_count"].plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Number of Capital Bikeshare Stations by Region")
plt.xlabel("Region")
plt.ylabel("Number of Stations")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

```
